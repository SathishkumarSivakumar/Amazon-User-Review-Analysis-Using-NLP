```python
import matplotlib.pyplot as plt

plt.plot(a["area"], a["order_total"])

plt.title("Area vs Order Total")
plt.xlabel("Area")
plt.ylabel("Order Total")

plt.show()
```

```python
import matplotlib.pyplot as plt

plt.plot(a["area"].head(10), a["order_total"].head(10))

plt.title("Area vs Order Total")
plt.xlabel("Area")
plt.ylabel("Order Total")
plt.grid()
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.plot(a["area"].head(10), a["order_total"].head(10),marker="*")

plt.title("Area vs Order Total")
plt.xlabel("Area")
plt.ylabel("Order Total")
plt.grid()
plt.show()
```

```python
abc=a.groupby("area")["order_total"].sum()
abc
```

```python
plt.bar(abc.index,abc.values)
plt.show()
```
